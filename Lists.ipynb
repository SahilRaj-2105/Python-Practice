```python
# . Given numbers = [1, 2, 2, 3, 4, 2, 5, 2], count how many times the number 2 appears without using the count() method.
l10 = [1, 2, 2, 3, 4, 2, 5, 2]
count=0
for i in l10:
    if i==2:
        count+=1
print(l10)
print(count)
```
