```python
# Q5: Take a student's marks out of 100 and print the grade using these rules: 90–100: A, 80–89: B, 70–79: C, 60–69: D, below 60: F. Print an error for marks outside 0–100.
x= int(input("Please enter the marks out of 100: "))
if 100>=x>=90:
    print("A")
elif 89>=x>=80:
    print("B")
elif 79>=x>=70:
    print("C")
elif 69>=x>=60:
    print("D")
elif 60>x>0:
    print("F")
else:
    print("Erro Marks outside Range")
```

```python
# Q6: Take a year and determine whether it is a leap year. A leap year is divisible by 400, or divisible by 4 but not by 100.

x=int(input())
if x%400 == 0 or (x%4==0 and x%100!=0):
    print("Year is leap")
else:
    print("not a leap year")
```

```python
# Q7: Create a simple calculator. Take two numbers and an operator (+, -, *, /) from the user. Perform the selected operation and handle division by zero and invalid operators.
a= int(input())
b= int(input())
opt= input("enter operator +,-,/,*")
if opt=="+":
    print(a+b)
elif opt=="-":
    print(a-b)
elif opt=="/":
    if b==0:
        print("Entered denominator is 0")
    else:
        print(a/b)
elif opt=="*":
    print(a*b)
else:
    print(":Entered wrong operator")
```
