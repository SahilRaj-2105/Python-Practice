```python
''' Q1: Take a positive integer from the user. Using loops and conditional statements, calculate:
- The sum of its digits.
- The number of even digits and odd digits.
- The largest digit in the number.
- Whether the number is a palindrome.'''

x= int(input())
y=x
sum=0
eve=0
odd=0
largest=0
rev_num=0
while y>0:
    z=y%10
# largest    
    if z>=largest:
        largest=z
# even odd        
    if z%2==0:
        eve+=1
    else:
        odd+=1
# sum        
    sum+= z

#rev_num
    rev_num= (rev_num*10)+z
    
    y//=10
    
print(sum)
print(f"No of even digits {eve} and No. of Odd digits {odd}")
print("Largest No: ", largest)
if rev_num==x:
    print("Palindrome")
else:
    print("Not a Palindrome")


```

```python
''' Q2: Take two integers, start and end, from the user. Print all prime numbers within that range, inclusive. Also print the total number of primes found and their sum.
Handle cases where start is less than 2.'''

x=int(input("enter the start: "))
y=int(input("enter the endpoint: "))

is_prime=0
sum=0

if x <=2:
    x=int(input("please enter a number greater tha 2: "))

for i in range(x,y+1):
    prime=True
    for j in range(2,i):
        if i%j==0:
            prime=False
            break
    if prime:
        is_prime+=1
        sum+=i
        print(i, end=" ")
print("/nSum of all the prime numbers between the range: ", sum)
print("total number if prime numbers in the range: ", is_prime)
         
        
```

```python
''' Take two integers representing a range. Print every Armstrong number in that range.
An Armstrong number equals the sum of its digits, each raised to the power of the number of digits. For example, \(153 = 1^3 + 5^3 + 3^3\).
Do not convert the numbers into strings.'''

def total_digit(n):
    if n==0:
        return 1
    total=0
    while n>0:
        total+=1
        n //=10
    return total

x=int(input())
y=int(input())

for i in range(x,y+1):
    temp=i
    z=total_digit(i)
    k=0
    while i>0:
        k+= (i%10)**z
        i//=10
    if k==temp:
        print(k)

    
```
