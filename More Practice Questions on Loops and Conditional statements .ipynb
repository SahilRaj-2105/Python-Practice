```python
''' Q1: Take a positive integer from the user. Using loops and conditional statements, calculate:
- The sum of its digits.
- The number of even digits and odd digits.
- The largest digit in the number.
- Whether the number is a palindrome.'''

x= int(input())
y=x
sum=0
eve=0
odd=0
largest=0
rev_num=0
while y>0:
    z=y%10
# largest    
    if z>=largest:
        largest=z
# even odd        
    if z%2==0:
        eve+=1
    else:
        odd+=1
# sum        
    sum+= z

#rev_num
    rev_num= (rev_num*10)+z
    
    y//=10
    
print(sum)
print(f"No of even digits {eve} and No. of Odd digits {odd}")
print("Largest No: ", largest)
if rev_num==x:
    print("Palindrome")
else:
    print("Not a Palindrome")


```

```python
''' Q3: Take two integers representing a range. Print every Armstrong number in that range.
An Armstrong number equals the sum of its digits, each raised to the power of the number of digits. For example, \(153 = 1^3 + 5^3 + 3^3\).
Do not convert the numbers into strings.'''

def total_digit(n):
    if n==0:
        return 1
    total=0
    while n>0:
        total+=1
        n //=10
    return total

x=int(input())
y=int(input())

for i in range(x,y+1):
    temp=i
    z=total_digit(i)
    k=0
    while i>0:
        k+= (i%10)**z
        i//=10
    if k==temp:
        print(k)

    
```

```python
''' Q4: Take two positive integers and calculate their greatest common divisor (GCD) and least common multiple (LCM) using loops and conditional statements.
Do not use Python's built-in GCD functions or external libraries.  '''

x= int(input())
y= int(input())

a=x
b=y

while b!=0:
    r= a % b
    a=b
    b= r
hcf= a
lcm= (x*y) // hcf

print(hcf, lcm)
    
```

```python
''' Take a positive integer and repeatedly calculate the sum of its digits until only one digit remains. '''
def sum_digits(n):
    sum=0
    while n>0:
        sum += n%10
        n //= 10
    if sum>=10:
        return sum_digits(sum)
    else:
        return sum
x= int(input())
ans= sum_digits(x)
print(ans)
```

```python
''' Create a number-guessing game with a fixed secret number of your choice between 1 and 100.
Give the user a maximum of 7 attempts. After each incorrect guess, tell them whether the secret number is higher or lower. Stop immediately when they guess correctly, and print the number of attempts used. If they fail, reveal the secret number. '''

x=7
ans=73
while x:
    g=int(input(f"Please enter your guess you have {x} attempts left: "))
    
    if g>73:
        print("Clue: The Secret number is a lower than your guess.")
    elif g<73:
        print("Clue: The Secret number is a Higher than your guess.")
    elif g==ans:
        print(f"Congratulations! You found the Secret Number and its {ans}")
        break
    x-=1
    
```

```python
'''Take an integer n from the user and print a hollow square of size n using nested loops.'''

n =int(input())

for i in range(1,n+1):
    for j in range(1,n+1):
        if i==1 or i==n or j==1 or j==n:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print("")
            
```

```python
''' Take an integer n from the user. Print an n × n multiplication grid, but print X instead of any product divisible by 3 or 5.'''
n =int(input())

for i in range(1,n+1):
    for j in range(1,n+1):
        if (i*j)%3==0 or (i*j)%5==0:
            print("X", end=" ")
        else:
            print(i*j, end=" ")
    print()
```

```python
''' Take a positive integer and find its second-largest distinct digit without converting it into a string. '''

n= int(input())
largest=0
sec_lar=0
while n>0:
    k= n%10
    if k>largest:
        sec_lar=largest
        largest=k
    elif k<largest and k>sec_lar:
        sec_lar=k
    n//=10
print(sec_lar)
    
```
