```python
#Q1: Write a program using a for loop to print all numbers from 1 to 20.
for i in range(21):
    print(i)
```

```python
# Q2:Write a program using a while loop to print all even numbers from 2 to 50.
x=2

while x<=50:
    if x%2==0:
        print(x)
    x+=1
    
```

```python
# Q3: Take a number from the user and print its multiplication table from 1 to 10.

x= int(input())
for i in range (1,11):
    print(i*x)
```

```python
# Q4:Take a positive integer from the user and calculate the sum of all numbers from 1 to that integer using a loop.

x=int(input())
total=0
for i in range(x+1):
    total+=i
print(total)    
```

```python
# Q5: Take a positive integer and calculate its factorial using a loop. For example, the factorial of 5 is 120.
x=int(input())
fac=1
for i in range(1,x+1):
    fac*=i
print(fac)
```

```python
# Q6: Take a positive integer and count how many digits it contains using a while loop. Do not convert the number into a string.
x=int(input())
ans=0
while x>0:
    ans+=1
    x//=10
print(ans)
    
```

```python
#Q7:Print all numbers from 1 to 100, but skip numbers divisible by 3 using continue. Stop the loop entirely when it reaches the first number divisible by 17 using break.
for i in range(1,100):
    if i%3==0:
        continue
    elif i%17==0:
        break
    else:
        print(i)
        
```

```python
#Q8: Take a positive integer and determine whether it is a prime number using a loop. Print the result.
x=int(input())
ans=True
for i in range(2,x):
    if x%i==0:
        ans=False
        break
print(ans)
```

```python
# Q9: Use nested loops to print the following pattern for 5 rows:
for i in range(1,6):
    for j in range(i):
        print("*", end="")
    print("\n")
    
```

```python
#Q10: Use nested loops to print a multiplication grid from 1 to 5, where each row displays the multiples of its row number from 1 to 5. Format the output neatly.
for i in range(1,6):
    for j in range(1,6):
        print(f"{i * j:2}", end=" ")
    print("\n")
    
```
