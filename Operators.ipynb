```python
# Q1:Take two numbers from the user and print their sum, difference, product, division, floor division, remainder, and exponentiation.

x=int(input())
y=int(input())

print("Sum: ", x+y)
print("Difference: ", x-y)
print("Product: ", x*y)
print("Division: ", x/y)
print("Floor Division: ", x//y)
print("Remainder: ", x%y)
print("Exponentiation: ", x**y)
```
